```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load dataset
df = pd.read_csv("../data/raw/updated_pollution_dataset.csv")

# 1. Check missing values
print("Missing values:")
print(df.isnull().sum())

# 2. Encode target column
label_encoder = LabelEncoder()
df["Air Quality"] = label_encoder.fit_transform(df["Air Quality"])

print("Encoded classes:")
for i, class_name in enumerate(label_encoder.classes_):
    print(i, "=", class_name)

# 3. Separate input and output features
X = df.drop("Air Quality", axis=1)
y = df["Air Quality"]

# 4. Scale input features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Convert scaled data back to dataframe
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# 6. Combine scaled input features with target column
processed_df = X_scaled_df.copy()
processed_df["Air Quality"] = y.values

# 7. Save processed dataset
processed_df.to_csv("../data/processed/cleaned_air_quality.csv", index=False)

processed_df.head()
```

```python
#data split
from sklearn.model_selection import train_test_split

X = processed_df.drop("Air Quality", axis=1)
y = processed_df["Air Quality"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
